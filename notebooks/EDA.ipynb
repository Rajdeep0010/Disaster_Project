# Objective 1 — Understanding Target Variables

---

## Purpose
Before building any model, we must understand what we are predicting.
Our 3 target variables are:
- `Total Deaths` — human fatalities per disaster event
- `Total Affected` — total people affected (injured + homeless + requiring assistance)
- `Total Damage ('000 US$)` — economic damage in thousands of USD

---

## Insight 1 — Data Availability (Missing Values)

| Target Column | Rows Filled | Total Rows | % Missing |
|---|---|---|---|
| Total Deaths | 740 | 804 | ~8% |
| Total Affected | 509 | 804 | ~37% |
| Total Damage (USD) | 202 | 804 | ~75% |

**Key Takeaway:**
> `Total Deaths` is the most reliable target with only 8% missing values.
> `Total Damage` is missing in **75% of rows** — severely limiting its use as a prediction target.
> Model training will be strongest for `Total Deaths`, weakest for `Total Damage`.

---

## Insight 2 — Extreme Right Skew in All 3 Targets

| Metric | Total Deaths | Total Affected | Total Damage (USD) |
|---|---|---|---|
| **Mean** | 12,361 | 49,50,834 | 7,45,123 |
| **Median (50%)** | 62 | 2,000 | 26,500 |
| **75th Percentile** | 203 | 9,00,000 | 5,39,300 |
| **Maximum** | 20,00,000 | 33,00,00,000 | 1,60,00,000 |
| **Mean/Median Gap** | **~200x** | **~2,475x** | **~28x** |

**Key Takeaway:**
> All 3 targets show severe right skew.
> The **median of 62 deaths** tells the true story — 50% of all disasters
> killed fewer than 62 people. The mean of 12,361 is being pulled up
> by a handful of catastrophic outliers (tsunamis, major epidemics, super cyclones).

---

## Insight 3 — Extreme Outliers

| Target | Max Value | Likely Event |
|---|---|---|
| Total Deaths | 20,00,000 | 2004 Indian Ocean Tsunami / Major Epidemic |
| Total Affected | 33,00,00,000 | Major river basin flood |
| Total Damage | 1,60,00,000 (USD '000) | Catastrophic multi-state disaster |

**Key Takeaway:**
> These single extreme events distort the entire distribution.
> A model trained on raw values will perform well only on catastrophic events
> and poorly on the vast majority of small-to-medium disasters.

---

## Action Items for Data Cleaning & Modelling

| Problem Identified | Solution in Phase 3 |
|---|---|
| Heavy right skew in all 3 targets | Apply **log1p transformation** before model training |
| Total Damage 75% missing | Use as **secondary target** only; report metrics separately |
| Extreme outliers at max values | Investigate top 5 events; decide whether to cap or flag |
| Total Affected 37% missing | Impute or restrict training rows to non-null values |

---

## Core Insight

> The data is dominated by small, contained disasters.
> A tiny number of catastrophic events are statistical giants that will
> mislead any ML model trained on raw values.
> **Log transformation is not optional — it is essential.**

## **Overview of Disaster Statistics**

The dataset provides a comprehensive look at various natural and biological disasters, categorized by type, frequency, seasonality, and measurement scales.

### **1. Most Frequent Disaster Types**

Flooding and atmospheric events dominate the records. **Floods** are the most common disaster type, occurring **333 times**, followed by **Storms (222)** and **Epidemics (69)**.

When breaking down these categories into subtypes:

* **Flood (General)** and **Riverine floods** are the most prevalent subtypes (148 and 144 occurrences, respectively).
* **Tropical cyclones** represent the largest portion of the "Storm" category with 122 recorded events.
* Other significant subtypes include **Landslides (wet)** (51) and **Flash floods** (37).

### **2. Seasonal Trends (Start Month)**

The occurrence of disasters shows a clear seasonal peak during the middle of the year.

* **Peak Period:** Disasters significantly increase starting in May (85) and reach their absolute peak in **August (109)** and **July (105)**.
* **Low Period:** February recorded the lowest frequency with only 14 events.
* **Trend:** There is a steady climb in frequency from April through August, likely corresponding with monsoon or cyclone seasons in the affected regions.

---

### **3. Magnitude and Measurement Scales**

The dataset uses different metrics to measure the intensity of these events:

* **Area ($Km^2$):** The most common measurement scale (353 uses), primarily used for floods and wildfires.
* **Wind Speed (Kph):** Used 222 times, corresponding to the number of storm events.
* **Temperature (°C):** Recorded 64 times, matching the "Extreme temperature" disaster count.
* **Seismic Activity (Moment Magnitude):** Used for the 28 recorded earthquakes.
* **Vaccination:** A metric used 45 times, likely in response to the recorded epidemic events.

### **4. River Basin Impact**

Data regarding specific river basins is relatively sparse compared to the total number of floods (with 720 null values in this column). However, among the recorded data:

* The **Brahmaputra** river system (including its tributaries and various sections like the Jamuna and Padma) is the most frequently mentioned geographic feature, accounting for the majority of specific river-related flood entries.
* Other mentioned basins include the Mahanadi, Ganges, Yamuna, and Sabarmati.

> This compresses extreme values while preserving the order and relationships.

# 📌 Objective 2 — Feature Validation

## The 4-Question Framework Applied to Every Feature
```
1. Does it vary enough to teach the model something?
2. Is it available before the disaster peaks?
3. Does it have enough filled rows to be reliable?
4. Is there a better alternative already available?
```

---

## Finding 2.1 — Disaster Type & Subtype ✅ Keep

| Disaster Type | Count | % of Data |
|---|---|---|
| Flood | 333 | 41% |
| Storm | 222 | 28% |
| Epidemic | 69 | 9% |
| Extreme Temperature | 64 | 8% |
| Others | 116 | 14% |

> Good variation confirmed. Each type behaves differently in severity.
> Subtype adds critical granularity — Riverine Flood vs Flash Flood
> have very different severity profiles despite the same parent type.
>
> ⚠️ Imbalance note: Model will predict flood severity most reliably
> due to 333 training examples vs only 28 for earthquakes.

---

## Finding 2.2 — Start Month & Season ✅ Keep + Engineer Season

| Month | Events | Season |
|---|---|---|
| February | 14 | Winter — lowest |
| June–September | ~100/month | Monsoon — peak |
| August | 109 | Monsoon — absolute peak |

> 8x difference between February and August — a real pattern, not noise.
> The monsoon signal is crystal clear and directly connected to flood severity.
> Engineering `Season` converts month numbers into meaningful context
> the model can interpret more effectively.

---

## Finding 2.3 — Magnitude ⚠️ Keep With Caution

| Scale | Count | Disaster Type |
|---|---|---|
| Area (Km²) | 353 | Floods |
| Wind Speed (Kph) | 222 | Storms |
| Temperature (°C) | 64 | Extreme Temp |
| Moment Magnitude | 28 | Earthquakes |

> Magnitude means completely different things across disaster types.
> Cannot be used as a raw number alone.
> Must always be paired with `Magnitude Scale` in the model
> so context is preserved.

---

## Finding 2.4 — River Basin ❌ Rejected

| Metric | Value |
|---|---|
| Null Values | 720 out of 804 |
| % Missing | **89%** |
| Usable Rows | 84 |

> Training on 84 rows is too few for reliable pattern learning.
> Geography is partially covered by `Primary_State` after cleaning Location.

---

## Finding 2.5 — Latitude & Longitude ❌ Model / 🗺️ Map Only

> 88% missing — unusable as model feature.

> Investigation revealed 3 categories:
>
> - Different locations same month → **Keep both** (genuinely separate events)
> - One filled location one empty → **Keep the filled row**
> - Both empty same type and year → **Keep higher data completeness row**
