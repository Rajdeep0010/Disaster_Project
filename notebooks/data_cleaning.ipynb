```python
df2.drop(columns=[
    'DisNo.', 'Classification Key', 'Disaster Group',
    'Historic', 'External IDs', 'Event Name',
    'OFDA/BHA Response', 'Appeal', 'Declaration',
    'ISO', 'Country', 'Region', 'Subregion',
    'Origin', 'Associated Types',
    "AID Contribution ('000 US$)",
    'No. Injured', 'No. Affected', 'No. Homeless',
    "Reconstruction Costs ('000 US$)",
    "Reconstruction Costs, Adjusted ('000 US$)",
    "Insured Damage ('000 US$)",
    "Insured Damage, Adjusted ('000 US$)",
    'CPI', 'Admin Units', 'GADM Admin Units',
    'Entry Date', 'Last Update',
    'Latitude', 'Longitude', 'River Basin'
], inplace=True)

print(df2.shape)
print(df2.columns.tolist())
```

```python
df2 = df2.sort_values(
    by=['Disaster Type','Start Year',
        'Start Month','Location'],
    na_position='last'
)

df2 = df2.drop_duplicates(
    subset=['Disaster Type','Start Year','Start Month'],
    keep='first'
)

print(df2.shape)
```

```python
df2['Start Day'] = df2['Start Day'].fillna(1)
df2['End Day'] = df2['End Day'].fillna(1)
df2['End Month'] = df2['End Month'].fillna(
    df2['Start Month'])
df2['End Year'] = df2['End Year'].fillna(
    df2['Start Year'])

df2['Start_Date'] = pd.to_datetime({
    'year'  : df2['Start Year'],
    'month' : df2['Start Month'],
    'day'   : df2['Start Day']
}, errors='coerce')

df2['End_Date'] = pd.to_datetime({
    'year'  : df2['End Year'],
    'month' : df2['End Month'],
    'day'   : df2['End Day']
}, errors='coerce')

print(df2[['Start_Date','End_Date']].head())
```

```python
df2['Duration'] = (
    df2['End_Date'] - df2['Start_Date']
).dt.days

df2['Duration'] = df2['Duration'].fillna(1)
df2['Duration'] = df2['Duration'].clip(lower=1)

print(df2['Duration'].describe())
```

```python
def get_season(month):
    if month in [6, 7, 8, 9]:
        return 'Monsoon'
    elif month in [10, 11]:
        return 'Post-Monsoon'
    elif month in [12, 1, 2]:
        return 'Winter'
    else:
        return 'Summer'

df2['Season'] = df2['Start Month'].apply(get_season)
print(df2['Season'].value_counts())
```

```python
colonial_map = {
    'madras'    : 'Tamil Nadu',
    'bombay'    : 'Maharashtra',
    'calcutta'  : 'West Bengal',
    'bengal'    : 'West Bengal',
    'orissa'    : 'Odisha',
    'mysore'    : 'Karnataka',
    'bezwada'   : 'Andhra Pradesh',
    'punjab province' : 'Punjab',
    'rajputana' : 'Rajasthan',
    'saurashtra': 'Gujarat',
    'kutch'     : 'Gujarat',
}

states = [
    'Andhra Pradesh', 'Arunachal Pradesh', 'Assam', 'Bihar',
    'Chhattisgarh', 'Goa', 'Gujarat', 'Haryana',
    'Himachal Pradesh', 'Jharkhand', 'Karnataka', 'Kerala',
    'Madhya Pradesh', 'Maharashtra', 'Manipur', 'Meghalaya',
    'Mizoram', 'Nagaland', 'Odisha', 'Punjab', 'Rajasthan',
    'Sikkim', 'Tamil Nadu', 'Telangana', 'Tripura',
    'Uttar Pradesh', 'Uttarakhand', 'West Bengal',
    'Delhi', 'Jammu', 'Kashmir'
]

def extract_state(location):
    if pd.isna(location):
        return 'Unknown'
    loc_lower = location.lower()
    for old, new in colonial_map.items():
        if old in loc_lower:
            return new
    for state in states:
        if state.lower() in loc_lower:
            return state
    return 'Unknown'

df2['Primary_State'] = df2['Location'].apply(
    extract_state)
print(df2['Primary_State'].value_counts().head(15))
```

```python
df2['Total Deaths'] = df2[
    'Total Deaths'].fillna(
    df2['Total Deaths'].median())

df2['Total Affected'] = df2[
    'Total Affected'].fillna(
    df2['Total Affected'].median())

df2["Total Damage ('000 US$)"] = df2[
    "Total Damage ('000 US$)"].fillna(
    df2["Total Damage ('000 US$)"].median())

df2['Start Month'] = df2[
    'Start Month'].fillna(
    df2['Start Month'].mode()[0])

print(df2.isnull().sum())
```

```python
#LOG TRANSFORMATION ON TARGETS
df2['Deaths_log'] = np.log1p(
    df2['Total Deaths'])

df2['Affected_log'] = np.log1p(
    df2['Total Affected'])

df2['Damage_log'] = np.log1p(
    df2["Total Damage ('000 US$)"])

print(df2[['Deaths_log',
                'Affected_log',
                'Damage_log']].describe())
```

```python
#Encoding categorical values
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in ['Disaster Type', 'Disaster Subtype',
            'Disaster Subgroup', 'Season',
            'Primary_State', 'Magnitude Scale']:
    df2[col + '_encoded'] = le.fit_transform(
        df2[col].astype(str))

print(df2.shape)
```

```python
#final check
print("Final shape:", df2.shape)
print("\nNull values remaining:")
print(df2.isnull().sum()[
    df2.isnull().sum() > 0])

df2.to_csv(
    r'C:\Users\boser\Desktop\DisasterProject\Dataset\df_cleaned.csv',
    index=False)

print("\nCleaned dataset saved successfully")
```
